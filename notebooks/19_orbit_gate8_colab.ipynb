```python
# Cell 2: Verify Checkpoint Hashes & Provenance
import hashlib
from pathlib import Path

def verify_sha256(path, expected):
    h = hashlib.sha256()
    with open(path, 'rb') as f:
        while chunk := f.read(1024*1024):
            h.update(chunk)
    digest = h.hexdigest()
    assert digest == expected, f'Mismatch: {digest} != {expected}'
    print(f'✓ Verified {path.name}: {digest}')

verify_sha256(Path('results/gate7/policy_gate7.pt'), '08615c72e2c2ec0c5e52e32e2b88cfeac2a51557b10dcdcd27daf0d57b9a44d8')
verify_sha256(Path('results/phase3/world_model_phase3.pt'), '918fdeee450d87de50cd1f7f796e69fb8446db87e57613a4948302daf3383306')
```
