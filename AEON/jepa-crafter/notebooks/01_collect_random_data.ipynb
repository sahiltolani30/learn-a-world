# Phase 1 — Random Experience Collection & Replay Buffer Verification

Collects initial exploratory trajectories with uniform random opaque tokens into the circular `ReplayBuffer`.

```python
import sys
sys.path.append('..')

from src.env.crafter_adapter import CrafterAdapter
from src.data.replay_buffer import ReplayBuffer
from src.data.sequence_sampler import SequenceSampler
from src.exploration.action_selector import ActionSelector
from src.training.collect import ExperienceCollector

adapter = CrafterAdapter(seed=123)
replay = ReplayBuffer(capacity=5000)
action_sel = ActionSelector()
collector = ExperienceCollector(adapter, replay, action_sel)

collector.reset(seed=123)
print("Collecting 500 random transitions...")
for _ in range(500):
    info = collector.step(mode="warmup")
    if info["done"]:
        collector.reset()

print(f"Replay buffer size: {replay.size} transitions.")
sampler = SequenceSampler(replay, context_length=4, prediction_horizon=8)
batch = sampler.sample_batch(batch_size=8)
print(f"Sampled batch context shape: {batch['obs_context'].shape}")
print(f"Sampled batch target_8 shape: {batch['target_8'].shape}")
adapter.close()
```
